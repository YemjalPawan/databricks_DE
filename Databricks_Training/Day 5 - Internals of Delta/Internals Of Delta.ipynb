```sql
--This will merge all parquet files into one parquet file upto 1GB if size is more then it will create another parquet file
--After otimizing the table 1 parquet file will be created upto 1 GB (by joining all small parquet files) but that small parquet files will not be deleted. It stored along with new parquet file 
optimize dev.demo.emp_eclerx
zorder by id
```

```sql
--while storing all parquet file it will create cost on storage
--so to remove the unused parquet file we can use vacuum command
--after vacuum command all unused parquet file will be deleted permanently but after 7 days or 168 hours 
--till that time we can time travel the table

vacuum dev.demo.emp_eclerx 
```

```sql
--if we want to delete now permanently then we can use retain commmand
--it will delete all unused parquet file permanently
-- we cant do time travel after executing below command

vacuum dev.demo.emp_eclerx retain 0 hours
```
