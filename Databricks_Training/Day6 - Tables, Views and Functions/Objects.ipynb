```sql
Types of Tables

1. Managed Tables
(default storage Account)
(Metadata managed by databricks)
(data(parquet) + metadata(crc,json))  -- stored in default storage account
In case of DROP : deletes everything (data + metadata)

    1. create table tbl_name (id int)
    2. create table tbl_name as select * from tbl_name
    3. spark.read.csv("path").write.saveAsTable("tbl_name")


2. External Tables
(data(parquet) + metadata(crc,json)) --stored in user location account
In case of Drop : in Catalog table will be deleted and (data + metadata) is untouched in user location

    1. create table tbl_name (id int) location "S3/ADLS"
    2. create table tbl_name location 'S3/ADLS' as select * from tbl_name
    3. spark.read.option('path','S3/ADLS').csv("path").write.saveAsTable("tbl_name")
```
