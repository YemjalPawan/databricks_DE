```sql
Views:
A virtual table
No Storage

standard view : saved in catalog and schema
temp view : not saved in catalog and schema (only valid till the spark session is active)
```

```sql
-- create view 
select * from dev.demo.transaction
```

```sql

Create view dev.demo.account_type_view as
select AccountType,sum(AmountUSD) as sum_amt from dev.demo.transaction
group by AccountType
order by sum_amt desc
limit 3
```

```sql
show views
```

```sql
select * from dev.demo.account_type_view
```
